```python
import pandas as pd
```

```python
morrowplots_df = pd.read_csv('data/morrowplots.csv')
morrowplots_df.head()
```

```python
mpcorn_df = morrowplots_df[morrowplots_df["corn"] == True]
mpcorn_df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))

sns.lineplot(data=mpcorn_df, x="year", y="yield_bush", hue="rotation", style="treated", lw=2)

plt.title("Morrow Plots Corn Yield Trends in Bushels per Acre")
plt.xlabel("Year")
plt.ylabel("Yield (bushels per acre)")
plt.legend(title="Rotation / Treated", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
```

I ran each of the code blocks above as per the instructions in Part 2. The data describes the year, plot, which rotation, what the yield was, what was applied, and a few other agronomic bits, like the population and planting date.

There are a few notes such as damage and details on how the treatments were applied. The median planting day was 126, which is May 6th.

The figure shows that overall, the yield went up a lot over time. You can see, however that the 1.0 rotation where treated is false (the solid pale line) has barely changed at all. That likely indicates that without rotation or fertility, no amount of modern hybrids, equipment, etc. are able to improve the corn yields. In fact, you can clearly see that the untreated lines of all rotations are below the treated ones. When you look at either treated or untreated lines, you can also see that they are consistently in the order 3, 2, 1 from top to bottom. That indicates that the crop rotation clearly improves yield. However, you can see that the difference is much less pronounced for the plots with fertility treatments than it is for the untreated plots. The data also has a huge amount of variability near the end. This is likely in part due to weather and damage making it so that the plots, regardless of treatment, were not able to reach the yield they potentially could have (a.k.a. bad years). I say that because it seems to be that for the years with big drops, all treatments drop to a similar level, which indicates that an outside factor is basically leveling the playing field and disrupting all treatments. Funnily enough, the only line that is barely affected by these events is the untreated monoculture corn, which never really yields high enough for it to matter.
